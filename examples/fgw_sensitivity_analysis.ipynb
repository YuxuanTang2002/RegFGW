# FGW parameter sensitivity

This notebook evaluates the sensitivity of the FGW distance to the FGW mixing parameter, $\alpha$, and the element embedding. Specifically, the Pearson ($r$) and Spearman ($\rho$) correlation coefficients of relative FGW distances ($\Delta d_{\mathrm{FGW}}$) with relative MACE-relaxed energies ($\Delta E_{\mathrm{MACE-relaxed}}$) and relative DFT-relaxed energies ($\Delta E_{\mathrm{relaxed}}$) are reported.


```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
from IPython.display import Markdown, display

# Input files
systems = ["KI_NaCl", "GaP_GaAs", "GaN_Al2O3"]
titles = ["KI/NaCl", "GaP/GaAs", "GaN/Al₂O₃"]
mace_paths = [f"mace_results/{system}.csv" for system in systems]
dft_paths = [f"dft_results/{system}.csv" for system in systems]
# Column names
alpha_columns = {
    0.25: "fgw_distance_megn_025",
    0.50: "fgw_distance_megn_050",
    0.75: "fgw_distance_megn_075",
}
embedding_columns = {
    "MEGNet": "fgw_distance_megn_050",
    "CGNF": "fgw_distance_cgnf_050",
    "Magpie": "fgw_distance_magp_050",
}
mace_energy_column = "mace_relaxed_energy",
dft_energy_column = "dft_relaxed_energy",
```

## Data processing and analysis

```python
alpha_records = []
embedding_records = []

for title, mace_path, dft_path in zip(titles, mace_paths, dft_paths):
    mace_df = pd.read_csv(mace_path)
    dft_df = pd.read_csv(dft_path)
    mace_relative_energy = mace_df["mace_relaxed_energy"]-mace_df["mace_relaxed_energy"].min()
    dft_relative_energy = dft_df["dft_relaxed_energy"]-dft_df["dft_relaxed_energy"].min()
    for alpha, fgw_column in alpha_columns.items():
        mace_relative_fgw = mace_df[fgw_column]-mace_df[fgw_column].min()
        dft_relative_fgw = dft_df[fgw_column]-dft_df[fgw_column].min()
        mace_r = float(pearsonr(mace_relative_fgw, mace_relative_energy).statistic)
        mace_rho = float(spearmanr(mace_relative_fgw, mace_relative_energy).statistic)
        dft_r = float(pearsonr(dft_relative_fgw, dft_relative_energy).statistic)
        dft_rho = float(spearmanr(dft_relative_fgw, dft_relative_energy).statistic)
        alpha_records.append(
            {
                "System": title,
                r"$\alpha$": f"{alpha:.2f}",
                r"$r(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{MACE-relaxed}})$": f"{mace_r:.3f}",
                r"$\rho(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{MACE-relaxed}})$": f"{mace_rho:.3f}",
                r"$r(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{relaxed}})$": f"{dft_r:.3f}",
                r"$\rho(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{relaxed}})$": f"{dft_rho:.3f}",
            }
        )
    for embedding, fgw_column in embedding_columns.items():
        mace_relative_fgw = mace_df[fgw_column]-mace_df[fgw_column].min()
        dft_relative_fgw = dft_df[fgw_column]-dft_df[fgw_column].min()
        mace_r = float(pearsonr(mace_relative_fgw, mace_relative_energy).statistic)
        mace_rho = float(spearmanr(mace_relative_fgw, mace_relative_energy).statistic)
        dft_r = float(pearsonr(dft_relative_fgw, dft_relative_energy).statistic)
        dft_rho = float(spearmanr(dft_relative_fgw, dft_relative_energy).statistic)
        embedding_records.append(
            {
                "System": title,
                "Embedding": embedding,
                r"$r(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{MACE-relaxed}})$": f"{mace_r:.3f}",
                r"$\rho(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{MACE-relaxed}})$": f"{mace_rho:.3f}",
                r"$r(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{relaxed}})$": f"{dft_r:.3f}",
                r"$\rho(\Delta d_{\mathrm{FGW}}, \Delta E_{\mathrm{relaxed}})$": f"{dft_rho:.3f}",
            }
        )

alpha_results = pd.DataFrame(alpha_records)
embedding_results = pd.DataFrame(embedding_records)
```
